# Sectional Curvature on SU(3)

## Background

For the bi-invariant metric $g(A,B) = -\mathrm{Tr}(AB)$ on skew-Hermitian matrices, the sectional curvature of the plane of two orthonormal $X, Y \in \mathfrak{u}(D)$ takes the closed form (Milnor, 1976)

$$
K(X,Y) = \frac{1}{4}\|[X,Y]\|^2 \geq 0 .
$$

It vanishes for commuting generators and grows with the failure of the two directions to commute. On $SU(2)$ it equals $\tfrac12$ for every plane, the value $1/R^2$ of the three-sphere of radius $\sqrt2$. This exercise compares planes of different curvature in $SU(3)$.

## Exercise

With $E_{jk} = |j\rangle\langle k|$ on $\mathbb{C}^3$, consider the generators

$$
X_1 = \tfrac{-i}{\sqrt2}\,\mathrm{diag}(1,-1,0), \quad
Y_1 = \tfrac{-i}{\sqrt6}\,\mathrm{diag}(1,1,-2), \quad
X_2 = \tfrac{-i}{\sqrt2}\,(E_{12}+E_{21}), \quad
Y_2 = \tfrac{1}{\sqrt2}\,(E_{21}-E_{12}), \quad
Y_3 = \tfrac{-i}{\sqrt2}\,(E_{23}+E_{32}).
$$

**(a)** Show that each generator lies in $\mathfrak{su}(3)$ and that each of the pairs $(X_1,Y_1)$, $(X_2,Y_2)$ and $(X_2,Y_3)$ is orthonormal.

**(b)** Compute $[X,Y]$ and $K(X,Y)$ for the three pairs.

**(c)** Two Hamiltonian evolutions leave the identity in the directions $X$ and $X+\epsilon Y$. For which pair do they stay parallel, and for which do they reconverge fastest? Relate the answer to the basis states the two generators act on.

## Solution

### Part (a)

All five matrices are skew-Hermitian and traceless. For the norms, $g(X_1,X_1) = \tfrac12\mathrm{Tr}\,\mathrm{diag}(1,1,0) = 1$, $g(Y_1,Y_1) = \tfrac16(1+1+4) = 1$, and $g(X_2,X_2) = g(Y_2,Y_2) = g(Y_3,Y_3) = \tfrac12\cdot 2 = 1$. The cross terms vanish, since $\mathrm{Tr}(X_1Y_1) \propto 1\cdot1 + (-1)\cdot1 = 0$, $X_2Y_2 = -\tfrac{i}{2}(E_{11}-E_{22})$ is traceless, and $X_2Y_3 = -\tfrac12 E_{13}$ has no diagonal entries.

### Part (b)

Diagonal matrices commute, so $[X_1,Y_1] = 0$ and $K = 0$. For the second pair $[X_2,Y_2] = -i\,\mathrm{diag}(1,-1,0)$, so $\|[X_2,Y_2]\|^2 = 2$ and $K = \tfrac12$, the value of $SU(2)$, since $X_2$ and $Y_2$ generate an $\mathfrak{su}(2)$ acting on $|1\rangle, |2\rangle$ only. For the third pair $[X_2,Y_3] = \tfrac12(E_{31}-E_{13})$, so $\|[X_2,Y_3]\|^2 = \tfrac12$ and $K = \tfrac18$.

### Part (c)

The diagonal pair commutes, the two evolutions generate a flat torus of diagonal unitaries, and the geodesics stay parallel. The pair acting on the same two states, $|1\rangle$ and $|2\rangle$, has the largest curvature and reconverges fastest. The pair $(X_2,Y_3)$ shares only the state $|2\rangle$, its commutator is smaller, and the curvature is a quarter of the same-block value.

---
## Numerical verification

Each pair is checked to be skew-Hermitian, traceless and orthonormal, and $K = \tfrac14\|[X,Y]\|^2$ is evaluated.

```python
import numpy as np

g = lambda A, B: np.real(np.trace(A.conj().T @ B))   # Hilbert-Schmidt metric, = -Tr(AB) on skew-Hermitian
comm = lambda A, B: A @ B - B @ A
E = lambda i, j: np.outer(np.eye(3)[i], np.eye(3)[j])

X1 = -1j*np.diag([1, -1, 0])/np.sqrt(2)
Y1 = -1j*np.diag([1, 1, -2])/np.sqrt(6)
X2 = -1j*(E(0,1) + E(1,0))/np.sqrt(2)
Y2 = (E(1,0) - E(0,1))/np.sqrt(2)
Y3 = -1j*(E(1,2) + E(2,1))/np.sqrt(2)

for name, (X, Y) in {"(X1,Y1)": (X1, Y1), "(X2,Y2)": (X2, Y2), "(X2,Y3)": (X2, Y3)}.items():
    assert np.allclose(X.conj().T, -X) and np.allclose(Y.conj().T, -Y)          # skew-Hermitian
    assert np.isclose(np.trace(X), 0) and np.isclose(np.trace(Y), 0)            # traceless
    assert np.isclose(g(X, X), 1) and np.isclose(g(Y, Y), 1) and np.isclose(g(X, Y), 0)
    C = comm(X, Y)
    K = g(C, C)/4
    print(f"{name}: |[X,Y]|^2 = {g(C,C):.4f},  K = {K:.4f}")

assert np.isclose(g(comm(X1,Y1), comm(X1,Y1))/4, 0)
assert np.isclose(g(comm(X2,Y2), comm(X2,Y2))/4, 1/2)
assert np.isclose(g(comm(X2,Y3), comm(X2,Y3))/4, 1/8)
print("K = 0, 1/2, 1/8 VERIFIED")
```

## Takeaway

The curvature of a plane in $\mathfrak{su}(3)$ measures how much the two generators overlap in the states they couple: $K=0$ for commuting generators, $\tfrac12$ within one $\mathfrak{su}(2)$ block, $\tfrac18$ for blocks sharing one state.
